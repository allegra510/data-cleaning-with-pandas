```python
flights_data = pd.read_csv('flights.csv')
flights_data.head(10)
```

```python
weather_data_pd = pd.read_csv('weather.csv')
weather_data_np = weather_data_pd.to_numpy()
```

```python
#Question 1 How many flights were there from JFK to SLC? Int
jfk_to_slc = flights_data[(flights_data['origin'] == 'JFK') & (flights_data['dest'] == 'SLC')]
q_1 = len(jfk_to_slc)
q_1
```

```python
#Question 2 How many airlines fly to SLC? Should be int
flights_to_slc = flights_data[flights_data['dest'] == 'SLC']
q_2 = flights_to_slc['carrier'].nunique()
q_2
    
```

```python
#Question 3 What is the average arrival delay for flights to RDU? float
avg_delay = flights_data[flights_data['dest'] == 'RDU']
q_3 = avg_delay['arr_delay'].mean()
q_3
```

```python
#Question 4 What proportion of flights to SEA come from the two NYC airports (LGA and JFK)? float
sea_flights = flights_data[flights_data['dest'] == 'SEA']
nyc_to_sea = sea_flights[sea_flights['origin'].isin(['LGA','JFK'])]
q_4 = len(nyc_to_sea) / len(sea_flights)
q_4
```

```python
#Question 5 Which date has the largest average depature delay? Pd slice with date and float. please make date a column. Preferred format is 2013/1/1 (y/m/d)
larg_dep_delay = flights_data.groupby(['year','month','day'])['dep_delay'].mean()
larg_dep_delay = larg_dep_delay.reset_index()
larg_dep_delay['date'] = (
    larg_dep_delay['year'].astype(str) + '/' +
    larg_dep_delay['month'].astype(str) + '/' +
    larg_dep_delay['day'].astype(str) 
)
q_5 = larg_dep_delay.sort_values(by='dep_delay', ascending = False).head(1)[['date','dep_delay']]
q_5
```

```python
#Question 6 Which date has the largest average arrival delay? pd slice with date and float
larg_arr_delay = flights_data.groupby(['year','month','day'])['arr_delay'].mean()
larg_arr_delay = larg_arr_delay.reset_index()
larg_arr_delay['date'] = (
    larg_arr_delay['year'].astype(str) + '/' +
    larg_arr_delay['month'].astype(str) + '/' +
    larg_arr_delay['day'].astype(str) 
)
q_6 = larg_arr_delay.sort_values(by='arr_delay', ascending = False).head(1)[['date','arr_delay']]
q_6
```

```python
# #Question 7 Which flight departing LGA or JFK in 2013 flew the fastest? pd slice with tailnumber and speed. speed = distance/airtime
fast_flight = flights_data[
    (flights_data['origin'].isin(['LGA','JFK'])) & 
    (flights_data['year']== 2013)
    ].copy()
fast_flight["speed"] = (
    fast_flight["distance"] / fast_flight["air_time"]
)
q_7 = fast_flight.sort_values(by='speed', ascending = False).head(1)[['tailnum','speed']]
q_7
```
